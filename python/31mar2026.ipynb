```python
#P
n=7
for i in range(1,n+1):
    if i==1 or i==n//2+1:
        print('*'*n)
    if i<=n//2:
        print('*'+' '*(n-2)+'*')
    else:
        print('*')
    
```

```python
#Q
j=n//2
for i in range(1,n+1):
    if i==1 or i==n:
        print('*'*(n+2))
    elif i>n//2+1 :
        print('*'+" "*j+'*'+' '*(n-j-1)+'*')
        j+=1
    else:
        print('*'+' '*(n)+'*')
        
    
```

```python
#R
j=n//2
for i in range(1,n+1):
    if i==1 or i==n//2:
        print('*'*n)
    elif i<n//2:
        print('*'+' '*(n-2)+'*')
    else:
        print('*'+' '*j+'*')
        j+=1
```

```python
#S
for i in range(1,n+1):
    if i==1 or i==n//2+1 or i==n:
        print('*'*n)
    elif i<=n//2:
        print('*')
        continue
    else:
        print(' '*(n-1)+'*')
```

```python
#V
j=(n)*2-1
k=0
for i in range(1,n+1):
    print(' '*k+'*'+" "*j+'*')
    j-=2
    k+=1
```

```python
#W
j=n//2
k=1
for i in range(1,n+1):
    if i<n//2+1:
        print('*'+' '*(n+2)+"*")
    else:
        print('*'+' '*j+'*'+' '*k+'*'+' '*j+'*')
        j-=1
        k+=2
    
```

```python
l=n
m=n//2+1
for i in range(1,n+1):
    if i<n//2+1:
        print(' '*i+'*'+' '*l+'*')
        l-=2
    else:
        print(' '*m+'*'+' '*l+'*')
        m-=1
        l+=2
```

```python
#Y
j=n-2
k=0
for i in range(1,n+1):
    if i==n//2+1:
        print(' '*k+'*')
    elif i<n//2+1:
        print(' '*k+'*'+" "*j+'*')
        j-=2
        k+=1
    else:
        print(' '*(n//2)+'*')
    
```
