# Factorial

```python
def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact*=i
    return fact
```

```python
print(factorial(9))
```

# Strong number


```python
def is_strong(num):
    temp=num
    Sum=0
    while num!=0:
        l=num%10
        Sum+=factorial(l)
        num//=10
    if temp==Sum:
        return True
```

```python
l=[]
for i in range(1,100000):
    if is_strong(i):
        l.append(i)
print(l)
        
```

# Armstrong number


```python
def is_armstrong(num):
    temp, temp1=num,num
    c,summ=0,0
    while temp!=0:
        c+=1
        temp//=10
    while temp1!=0:
        last=temp1%10
        summ+=last**c
        temp1//=10
    if num==summ:
        return True
```

```python
al=[]
for i in range(1,2000):
    if is_armstrong(i):
        al.append(i)
print(al)
```
