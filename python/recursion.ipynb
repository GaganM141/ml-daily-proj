# 1. Write a program for factorial .

```python
def fact(n):
    if n==0 or n==1:
        return 1
    return n*fact(n-1)
print(f'factorial of number 18 is {fact(18)}')
```

```python
print(f'factorial of number 4 is {fact(4)}')
```

```python
def int_rev(n, i=0):
    print(i, end=' ')
    if i==n:
        return
    int_rev(n,i=i+1)
    print(i, end=' ')
int_rev(23)
```

# Reverse a String

```python
def rev_str(s,r,i):
    if i==-1:
        return r
    r=r+s[i]
    return rev_str(s,r,i=i-1)
r=''

print(rev_str('hello',r,len('hello')-1))
```
