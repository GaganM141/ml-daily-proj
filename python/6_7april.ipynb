# 1. Write a program to convert binary to decimal

```python
a=int(input('enter a binary number:'))
dec=0
p=0
while a!=0:
    l=a%10
    dec=dec+l*2**p
    a//=10
    p+=1
print(dec)
```

# 2. Write a program to convert decimal to binary using functions

```python
def d2b(d):
    b,rem='',0
    while d>0:
        rem=d%2
        b=str(rem)+b
        d//=2
    b=int(b)
    return b
binary=d2b(7)
print(binary)
```

```python
def extract_nd(e):
    out=[]
    for i in e:
        if bool(i)==True:
            out.append(i)
    return out
non_default=extract_nd([1,0,'hi',(),[1,2,3],0j])
print(non_default)
```

## o='b4a5c2'

```python
def count(i):
    d={}
    for j in i:
        if j not in d:
            d[j]=1
        else:
            d[j]+=1
    s=''
    for k in d:
        s+=k
        s+=str(d[k])
        
    return s
out=count('lsalaj;ajfddfll')
print(out)
```

# 5. Write a function for fibanacci series

```python
def fib(n):
    l=[0,1]
    for i in range(2,n):
        l.append(l[i-2]+l[i-1])
    return l
fibo=fib(34)
print(fibo)
        
```
